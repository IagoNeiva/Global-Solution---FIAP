```python
import requests

def chamar_IA (prompt):
  payload = {
      "model": "llama3.2:1b",
      "messages": [
          {
              "role": "system",
              "content": """Você é o Mission Control AI.

  Sua função é analisar os indicadores da missão e gerar alertas para o comandante.

  Regras:
  - Respostas curtas.
  - Linguagem técnica e consultiva.
  - Utilizar tópicos.
  - Sempre quebrar linhas.
  - Não escrever textos longos.
  - Destacar apenas sistemas em ATENÇÃO ou CRÍTICO.
  - Informar impacto operacional.
  - Gerar recomendações práticas e objetivas.
  - Finalizar com o nível de prioridade.
      """

          },
          {
              "role": "assistant",
              "content": """ ALERTA DE MISSÃO

  Status Geral:
  • MISSÃO EM ATENÇÃO

  Sistemas Afetados:
  • Temperatura: 36°C → CRÍTICO
  • Comunicação: 42% → ATENÇÃO
  • Bateria: 38% → ATENÇÃO

  Impacto:
  • Risco de superaquecimento dos módulos.
  • Possível degradação do desempenho operacional.

  Recomendações:
  • Verificar imediatamente o sistema de controle térmico.
  • Priorizar economia de energia.
  • Monitorar estabilidade da comunicação.

  Prioridade:
  • ALTA"""
          },
          {
    "role": "assistant",
    "content": """
  ALERTA CRÍTICO

  Status Geral:
  • MISSÃO CRÍTICA

  Sistemas Afetados:
  • Comunicação: 28% → CRÍTICO
  • Bateria: 19% → CRÍTICO
  • Oxigênio: 78% → CRÍTICO
  • Estabilidade: 35% → CRÍTICO

  Impacto:
  • Elevado risco de perda de controle operacional.
  • Possibilidade de interrupção da missão.

  Recomendações:
  • Acionar protocolo de contingência.
  • Restabelecer comunicação com a base.
  • Ativar modo de economia máxima de energia.
  • Executar procedimentos de suporte à vida.
  • Suspender operações não essenciais.

  Prioridade:
  • MÁXIMA
  """
  },
          {
    "role": "user",
    "content": prompt
}

      ],
      "stream": False
  }

  response = requests.post(
      "http://localhost:11434/api/chat",
      json=payload
  )

  return response.json()["message"]["content"]
```

```python
def temperatura_interna(temperatura):
    if temperatura < 18:
        classificacao = 'ATENÇÃO | Risco de congelamento'
        pontuacao_risco = 1
    elif 18 <= temperatura < 30:
        classificacao = 'NORMAL | Temperatura estável'
        pontuacao_risco = 0
    elif 30 <= temperatura < 35:
        classificacao = 'ATENÇÃO | Temperatura elevada'
        pontuacao_risco = 1
    elif temperatura >= 35:
        classificacao = 'CRÍTICO | Risco de superaquecimento'
        pontuacao_risco = 2
    return classificacao, pontuacao_risco

def comunicacao_base(comunicacao):
    if comunicacao < 30:
        classificacao = 'CRÍTICO | Comunicação com a base em nivel crítico'
        pontuacao_risco = 2
    elif 30 <= comunicacao < 59:
        classificacao = 'ATENÇÃO | Comunicação instável'
        pontuacao_risco = 1
    elif comunicacao >= 60:
        classificacao = 'NORMAL | Comunicação estável'
        pontuacao_risco = 0
    return classificacao, pontuacao_risco

def sistema_energia(bateria):
    if bateria < 20:
        classificao = 'CRÍTICO | Bateria em nivel crítico'
        pontuacao_risco = 2
    elif 20 <= bateria < 49:
        classificao = 'ATENÇÃO | Bateria abaixo do recomendado'
        pontuacao_risco = 1
    elif bateria >= 50:
        classificao = 'NORMAL | Energia estável'
        pontuacao_risco = 0
    return classificao, pontuacao_risco

def suporte_oxigenio(oxigenio):
    if oxigenio < 80:
        classificacao = 'CRÍTICO | Oxigênio em nivel crítico'
        pontuacao_risco = 2
    elif 80 <= oxigenio < 89:
        classificacao = 'ATENÇÃO | Oxigênio abaixo do ideal'
        pontuacao_risco = 1
    elif oxigenio >= 90:
        classificacao = 'NORMAL | Oxigênio adequado'
        pontuacao_risco = 0
    return classificacao, pontuacao_risco

def estabilidade_operacional(estabilidade):
    if estabilidade < 40:
        classificacao = 'CRÍTICO | Estabilidade operacional crítica'
        pontuacao_risco = 2
    elif 40 <= estabilidade < 69:
        classificacao = 'ATENÇÃO | Estabilidade operacional reduzida'
        pontuacao_risco = 1
    elif estabilidade >= 70:
        classificacao = 'NORMAL | Estabilidade operacional adequada'
        pontuacao_risco = 0
    return classificacao, pontuacao_risco


def classificacao_risco(pontuacao_total):
    if 0 <= pontuacao_total <= 2:
        classificacao = 'MISSÃO ESTÁVEL'
    elif 3 <= pontuacao_total <= 7:
        classificacao = 'MISSÃO EM ATENÇÃO'
    elif pontuacao_total >= 8:
        classificacao = 'MISSÃO CRÍTICA'
    return classificacao


def tendencia_missao(risco_ultimo, risco_primeiro):
    if risco_ultimo > risco_primeiro:
        tendencia_missao = 'A missão apresentou tendência de piora'
    elif risco_ultimo < risco_primeiro:
        tendencia_missao = 'A missão apresentou tendência de melhora'
    elif risco_ultimo == risco_primeiro:
        tendencia_missao = 'A missão permaneceu estável'
    return tendencia_missao

def consultar_ia(temperatura, comunicacao, bateria, oxigenio, estabilidade):
    prompt = f"""
    Temperatura: {temperatura}
    Comunicação: {comunicacao}
    Bateria: {bateria}
    Oxigênio: {oxigenio}
    Estabilidade: {estabilidade}

    Gere um alerta operacional.
    """

    # chamada do Ollama
    resposta = chamar_IA(prompt)

    return resposta

soma_temperatura = 0
soma_comunicacao = 0
soma_energia = 0
soma_oxigenio = 0
soma_estabilidade = 0

lista_recomendacoes = []

dados_missao = [
    [24, 92, 88, 96, 90],
    [27, 80, 72, 94, 85],
    [31, 65, 58, 91, 70],
    [36, 42, 38, 87, 55],
    [39, 28, 19, 78, 35],
    [34, 55, 32, 82, 50]
]

# [temperatura, comunicacao, bateria, oxigenio, estabilidade]

risco_temperatura_acumulada = 0
risco_comunicacao_acumulada = 0
risco_energia_acumulada = 0
risco_oxigenio_acumulada = 0
risco_estabilidade_acumulada = 0


lista_riscos_ciclos = []

risco_total = 0
risco_ciclo = 0

# Cabeçalho do relátorio
print('=' * 40)
print('MISSION CONTROL AI')
print('=' * 40)
print(f'Missão: Orion Test Alpha\nEquipe: Equipe Apollo\nQuantidade de ciclos analisados: 6')
print('=' * 40)

# Relatorio de cada ciclo
for h, ciclo in enumerate(dados_missao):
    print(f'\nCICLO {h +1}')
    print('-' * 40)
    for i, valor in enumerate(ciclo):
        if i == 0: # Temperatura
            alerta, risco = temperatura_interna(valor)
            print(f'Temperatura: {valor} | {alerta}')
            risco_temperatura_acumulada += risco
            risco_total += risco
            risco_ciclo += risco
            soma_temperatura += valor
            if risco == 2:
                lista_recomendacoes.append('Alerta temperatura: Verificar controle térmico da missão.')
            print()

        elif i == 1: # Comunicação
            alerta, risco = comunicacao_base(valor)
            print(f'Comunicação: {valor} | {alerta}')
            risco_comunicacao_acumulada += risco
            risco_total += risco
            risco_ciclo += risco
            soma_comunicacao += valor
            if risco == 2:
                lista_recomendacoes.append('Alerta comunicação: Tentar restabelecer contato com a base.')
            print()

        elif i == 2: # Energia
            alerta, risco  = sistema_energia(valor)
            print(f'Bateria: {valor} | {alerta}')
            risco_energia_acumulada += risco
            risco_total += risco
            risco_ciclo += risco
            soma_energia += valor
            if risco == 2:
                lista_recomendacoes.append('Alerta bateria: Ativar modo de economia de energia.')
            print()

        elif i == 3: # Oxigênio
            alerta, risco = suporte_oxigenio(valor)
            print(f'Oxigênio: {valor} | {alerta}')
            risco_oxigenio_acumulada += risco
            risco_total += risco
            risco_ciclo += risco
            soma_oxigenio += valor
            if risco == 2:
                lista_recomendacoes.append('Alerta oxigênio: Acionar protocolo de suporte à vida.')
            print()

        elif i == 4: # Estabilidade
            alerta, risco = estabilidade_operacional(valor)
            print(f'Estabilidade: {valor} | {alerta}')
            risco_estabilidade_acumulada += risco
            risco_total += risco
            risco_ciclo += risco
            soma_estabilidade += valor
            print(f'\nPontuação de risco do ciclo: {risco_ciclo}')
            if risco == 2:
                lista_recomendacoes.append('Alerta estabilidade: Reduzir operações não essenciais.')
                # Relatorio do ciclo
            print(f'\nClassificação do ciclo {h+1}: {classificacao_risco(risco_total)}\n')
            temperatura, comunicacao, energia, oxigenio, estabilidade = ciclo
            while True:
              receber_analise = input('Deseja receber análise e recomendações da IA (sim | nao)? ').lower()
              if receber_analise == 'sim':
                print(consultar_ia(temperatura, comunicacao, energia, oxigenio, estabilidade))
                print('Preparando o próximo ciclo...')
                time.sleep(3)
                break
              elif receber_analise == 'nao' or receber_analise == 'não':
                print('Preparando o próximo ciclo...')
                time.sleep(3)
                break
              else:
                print('Resposta inválida.')


            print(f"{'-' * 40} \n")

    lista_riscos_ciclos.append(risco_ciclo)
    risco_ciclo = 0



# Relatório final
print('=' * 40)
print('RELATÓRIO FINAL DA MISSÃO')
print('=' * 40)
print(f'Missão: Orion Test Alpha\nEquipe: Equipe Apollo\nQuantidade de ciclos analisados: 6\n')

# Media dos valores
media_temperatura = soma_temperatura / 6
media_comunicacao = soma_comunicacao / 6
media_energia = soma_energia / 6
media_oxigenio = soma_oxigenio / 6
media_estabilidade = soma_estabilidade / 6
print(f'Média de temperatura: {media_temperatura:.2f}°C')
print(f'Média de comunicação: {media_comunicacao:.2f}%')
print(f'Média de bateria: {media_energia:.2f}%')
print(f'Média de oxigênio: {media_oxigenio:.2f}%')
print(f'Média de estabilidade: {media_estabilidade:.2f}%')

# Ciclo mais crítico
indice_maior = 0
maior = 0
for i, num in enumerate(lista_riscos_ciclos):
    if num > maior:
        maior = num
        indice_maior = i
print(f'\nCiclo mais crítico: Ciclo {indice_maior+1}')
print(f'Maior pontuação de risco: {maior}')


# Tendência da missão
print(f'\nTendência da missão:')
print(f'{tendencia_missao(lista_riscos_ciclos[5], lista_riscos_ciclos[0])}\n')

# Pontuação acumulada por área
print('Pontuação acumulada por área:')
riscos_area = {'Temperatura interna': risco_temperatura_acumulada,'Comunicação com a base': risco_comunicacao_acumulada,'Sistema de energia': risco_energia_acumulada,'Suporte de oxigênio': risco_oxigenio_acumulada,'Estabilidade operacional': risco_estabilidade_acumulada}
for chave, valor in riscos_area.items():
    print(f'{chave}: {valor} pontos')

# Área mais afetada
print('\nÁrea mais afetada:')
maior_risco = max(riscos_area)
print(maior_risco)

# Classificação final
print('\nClassificação final da missão:')
print(f'{classificacao_risco(risco_total)}')

# Conclusão
print('\nConclusão da missão:')
print('A missão apresentou instabilidade relevante durante a operação. Apesar da tentativa de recuperação no último ciclo, ainda existem sistemas em atenção e a equipe deve manter o plano de contingência ativo.')
```
